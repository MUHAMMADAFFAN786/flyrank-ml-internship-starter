## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

Comparison plan: I train the selected model on the training data and evaluate it using the same metric used for the Week-4 baseline. I will report both results in a small comparison table. The purpose is not simply to obtain a more complex model, but to determine whether the new method provides a meaningful improvement over the baseline.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.import json
import pandas as pd
import os

# Load the verified Week-4 baseline/model results
results_path = "outputs/model_results.json"

# Check if the outputs directory exists, create if not
if not os.path.exists("outputs"):
    os.makedirs("outputs")
    print("Created 'outputs' directory.")

# Check if the model_results.json file exists, create with dummy data if not
if not os.path.exists(results_path):
    dummy_results = {
      "baseline": {
        "baseline_precision_at_50": 0.1
      },
      "models": {
        "random_forest": {
          "precision_at_50": 0.15
        }
      },
      "split_strategy": "Time-aware split for fair evaluation"
    }
    with open(results_path, "w") as f:
        json.dump(dummy_results, f, indent=2)
    print(f"Created dummy '{results_path}' for demonstration purposes.")
    print("Please replace with actual model results if available.")


with open(results_path, "r") as f:
    results = json.load(f)

# Baseline and Random Forest results
baseline_score = results["baseline"]["baseline_precision_at_50"]
rf_score = results["models"]["random_forest"]["precision_at_50"]

# Comparison table
comparison = pd.DataFrame({
    "Method": [
        "Week-4 Hand-written Baseline",
        "Random Forest"
    ],
    "Precision@50": [
        baseline_score,
        rf_score
    ]
})

comparison["Precision@50"] = comparison["Precision@50"].round(3)

display(comparison)

print(f"Baseline Precision@50: {baseline_score:.3f}")
print(f"Random Forest Precision@50: {rf_score:.3f}")
print(f"Relative improvement: {rf_score / baseline_score:.1f}x")

print("\nValidation strategy:")
print(results["split_strategy"])
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

Error analysis: After evaluation, I will inspect the model's incorrect predictions and look for recurring patterns in the errors. I will consider whether the errors appear to come from weak or noisy features, overlapping classes, insufficient training data, or limitations of the chosen model. The analysis will focus on what the errors reveal about the problem rather than treating the score alone as the result.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Error / result interpretation check

baseline = baseline_score
model = rf_score

print("Error analysis summary")
print("-----------------------")
print(f"Baseline Precision@50: {baseline:.3f}")
print(f"Random Forest Precision@50: {model:.3f}")

if model > baseline:
    print("\nThe Random Forest improves Precision@50 over the baseline.")
    print("The improvement suggests that the learned features capture useful signal")
    print("that the simple baseline does not capture.")
elif model < baseline:
    print("\nThe Random Forest does not improve over the baseline.")
    print("This suggests that the added model complexity does not provide useful")
    print("additional signal under the current validation setup.")
else:
    print("\nThe Random Forest matches the baseline.")
    print("The added complexity does not provide a measurable improvement.")

print("\nImportant limitation:")
print("This comparison does not by itself prove generalization. The result should")
print("be interpreted together with the validation split and the underlying data.")
```
