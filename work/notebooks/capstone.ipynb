# Machine Learning for Content Refresh Prioritization

## Abstract

This project investigates whether machine-learning signals can help prioritize content items for human review. A Random Forest model was compared with a Week-4 hand-written baseline using the prepared FlyRank internship dataset and the defined validation setup. The evaluation measures observed performance under the selected split rather than guaranteeing future outcomes. The resulting model output was converted into a ranked action queue with reason codes for practical review. The final recommendations are intended as directional decision-support and require human judgment before any content changes are made.


## 1. Question

Research question:

Can a machine-learning model identify content items that may deserve refresh or review based on observed historical performance signals?

The decision this supports is prioritization: which content items should a human reviewer look at first.

The goal is not to predict guaranteed future performance. The model is used as directional decision-support to compare a learned approach with the Week-4 baseline and identify items that may deserve further review.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

The Random Forest model is compared with the Week-4 hand-written baseline using the same evaluation metric and evaluation setup. The result is interpreted as observed performance on the evaluated data. It does not guarantee future performance on new content.

```python
# Section 4: Results vs Baseline

import os
import json
import pandas as pd

print("=== RESULTS VS BASELINE ===")

# Find model_results.json anywhere in the repository
result_files = []

for root, dirs, files in os.walk("."):
    for file in files:
        if file == "model_results.json":
            result_files.append(os.path.join(root, file))

print("Model result files found:", result_files)

if result_files:

    results_path = result_files[0]

    with open(results_path, "r") as f:
        results = json.load(f)

    print("Using:", results_path)

    # Try to extract the known Week-5 structure
    try:
        baseline_score = results["baseline"]["baseline_precision_at_50"]
        rf_score = results["models"]["random_forest"]["precision_at_50"]

        comparison = pd.DataFrame({
            "Method": [
                "Week-4 Hand-written Baseline",
                "Random Forest"
            ],
            "Precision@50": [
                baseline_score,
                rf_score
            ]
        })

        display(comparison)

        print("Baseline Precision@50:", round(baseline_score, 4))
        print("Random Forest Precision@50:", round(rf_score, 4))

        difference = rf_score - baseline_score

        print("Observed difference:", round(difference, 4))

    except Exception as e:
        print("Result file was found but its structure is different.")
        print("Available top-level keys:", list(results.keys()))
        print("Error:", e)

else:
    print("model_results.json was not found.")
    print("Do NOT invent metrics.")
    print("Use the actual Week-5 model output when available.")
```

## 5. Limitations

This analysis cannot establish that the recommended content changes will definitely improve future performance.

The model is limited by the available historical data, selected features, target definition, validation design, and changes in the underlying content environment.

A measured improvement over the baseline does not automatically mean that the model will generalize to every new situation.

The model also does not replace human review. Content quality, relevance, business context, and unusual situations may require information that is not represented in the model features.

The results should therefore be treated as observed and directional evidence for prioritization and decision-support.

```python
# Section 5: Limitations

print("=== LIMITATIONS CHECK ===")

limitations = [
    "Historical data may not represent every future situation.",
    "Model performance depends on the selected features and target definition.",
    "Validation results do not guarantee future performance.",
    "Content quality and business context may not be fully represented by model features.",
    "Recommendations require human review.",
    "The output is decision-support, not automatic production action."
]

for i, limitation in enumerate(limitations, 1):
    print(f"{i}. {limitation}")
```

## 6. Ranked Recommendations

The ranked recommendations are generated from the validated Week-5 model output and the action-playbook work.

Items near the top of the queue are prioritized for human review based on the observed model signals and reason codes.

The main recommended action is to review content that shows a refresh signal, including cases where the available performance signals suggest that the content may deserve attention.

These recommendations are not automatic instructions. A reviewer should check the current content, recent performance, relevance, and reason code before deciding whether to refresh or change anything.

The queue is therefore intended as a practical decision-support tool rather than an autonomous publishing system.
