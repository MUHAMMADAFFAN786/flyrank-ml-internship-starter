## The Unit of Analysis

The unit of analysis is a single web page.

Each row in the dataframe represents one web page with search performance metrics such as search volume, competition, CTR, average position, and other content features. These values help estimate the SEO optimization priority for that page.

```python
from pathlib import Path
import os
import subprocess
import pandas as pd

repo_url = "https://github.com/MUHAMMADAFFAN786/flyrank-ml-internship-starter.git"
csv_relative_path = Path("data/raw/content_refresh_anonymized.csv")

candidate_paths = [
    Path.cwd(),
    Path("/content/flyrank-ml-internship-starter"),
]

repo_path = next((p for p in candidate_paths if (p / csv_relative_path).exists()), None)

if repo_path is None:
    repo_path = Path("/content/flyrank-ml-internship-starter")
    if not repo_path.exists():
        subprocess.run(["git", "clone", repo_url, str(repo_path)], check=True)
    elif not (repo_path / csv_relative_path).exists():
        subprocess.run(["git", "-C", str(repo_path), "pull", "--ff-only"], check=True)

os.chdir(repo_path)
df = pd.read_csv(csv_relative_path)

print("Repo ready:", os.getcwd())
print("Dataset Shape:", df.shape)
print("Columns:", list(df.columns))
display(df.head())
```

## Why ML Beats a Fixed Rule

A fixed rule cannot capture the complex relationship between impressions, clicks, CTR, and average position.

Machine learning can learn patterns from multiple features at the same time and rank pages more effectively than a simple if-else rule. This supports better decisions for content optimization.
